```python
import os
# Autodl通用绝对路径
root = "/root/autodl-tmp/agent_dataset/processed"

# 校验目录
print("目录是否存在：", os.path.exists(root))
print("目录内一级内容：", os.listdir(root))

# 递归遍历所有文件
all_files = []
for dirpath, _, filenames in os.walk(root):
    for fname in filenames:
        full_path = os.path.join(dirpath, fname)
        all_files.append(full_path)

print(f"\n一共找到 {len(all_files)} 个文件")
# 输出前20个预览
for path in all_files[:20]:
    print(path)
```
